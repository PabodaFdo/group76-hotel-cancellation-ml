```python
import pandas as pd
import numpy as np
from pathlib import Path

DATA_PATH = Path("../../data/raw/hotel_bookings.csv")
# If the repository copy is stored elsewhere, update DATA_PATH only.
df = pd.read_csv(DATA_PATH)

print("Shape:", df.shape)
df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nColumn names:")
print(df.columns.tolist())
```

```python
structure = pd.DataFrame({
    "dtype": df.dtypes.astype(str),
    "non_null": df.notna().sum(),
    "missing": df.isna().sum(),
    "missing_pct": (df.isna().mean() * 100).round(3),
    "unique": df.nunique(dropna=True)
})
structure
```

```python
categorical_columns = df.select_dtypes(include=["object", "category"]).columns.tolist()
numeric_columns = df.select_dtypes(include=["number"]).columns.tolist()

print("Categorical columns:", len(categorical_columns))
print(categorical_columns)
print("\nNumeric columns:", len(numeric_columns))
print(numeric_columns)
```

```python
# Inspect low-cardinality columns and the target.
for col in df.columns:
    if df[col].nunique(dropna=True) <= 15:
        print(f"\n--- {col} ---")
        print(df[col].value_counts(dropna=False))
```

```python
target = "is_canceled"

print("Target distribution:")
print(df[target].value_counts(dropna=False))
print("\nTarget proportion:")
print(df[target].value_counts(normalize=True, dropna=False).round(4))
```

```python
# Duplicate check for documentation only.
print("Exact duplicate rows:", df.duplicated().sum())
```

## Leakage check

For this project the prediction target is `is_canceled`.

`reservation_status` and `reservation_status_date` describe the final reservation outcome/status. Because these fields can contain information recorded after the cancellation decision, they are potential target leakage and should not be used as ordinary predictive features unless the group explicitly defines a prediction point where they are available.

This is a modelling decision to document before Phase 3/4.


```python
# Quick check of suspicious/post-outcome columns
for col in ["reservation_status", "reservation_status_date"]:
    print(col, "->", df[col].nunique(dropna=True), "unique values")
    print(df[col].value_counts().head())
```
